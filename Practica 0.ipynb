### 5. Podemos usar la función range() para crear listas: 

````python
a = list(range(4))
b = list(range(5,10))
c = list(range(0, 100, 10))
d = list(range(10, 0, -1))
print(a, b, c, d)
"""
En todos los casos recuerda que el extremo derecho no es inclusivo (range(5,10) genera una secuencia de
5 a 9, y range(10, 0, -1) genera una secuencia de 10 a 1). 
"""
````

### 6. Usamos también los corchetes para indexar elementos y porciones de una lista:

````python
a = list(range(1,100))
print(a[0], a[1], a[10], a[25], a[50])
print(a[:10])
print(a[90:])
print(a[25:50]) 
"""
Observa que cuando usamos : para extraer rangos, el extremo derecho tampoco es inclusivo. 
"""
````

### 8. Python cuenta con ciclos for y while:

````python
lista = [1,2,3]
contador = 0
for i in lista:
    print(i)
for i in range(0,11):
    print(i)
for i in range(10, -1, -1):
    print(i)
while contador <= 10:
    print(contador) 
    contador += 1
````

### 9. Podemos definir funciones con la palabra clave def:

````python
def cuadrado(n):
    return n*n
def factorial(n):
    if n <= 1:
        return n
    else:
        return n * factorial(n-1)
print(cuadrado(5))
print(factorial(5))
````
