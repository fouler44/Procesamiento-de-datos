### **Actividad 15. La naturaleza como optimizadora**

```python
class Poblacion:
    def __init__(self, individuos: list):
        self.individuos = individuos
        self.generacion = 1
        self.tamano = len(individuos)

    def mostrar_poblacion(self):
        print("Generación:", self.generacion)
        print("Tamaño de la población:", self.tamano)
        for individuo in self.individuos:
            print("-", individuo.nombre)

    def reproducir(self):
        print("Los individuos seleccionados se reproducen para formar nuevos descendientes.")

    def evolucionar(self):
        print("La población evoluciona a una nueva generación.")
        self.generacion += 1

    def evaluar_poblacion(self):
        print("Se evalúa la aptitud general de toda la población.")
        for individuo in self.individuos:
            individuo.evaluar_fitness()
```

```python
class Individuo:
    def __init__(self, nombre, genes):
        self.nombre = nombre
        self.genes = genes
        self.fenotipo = "Definido en la información genética"
        self.fitness = 0

    def mostrar_info(self):
        print("Nombre:", self.nombre)
        print("Genes:", self.genes)
        print("Fenotipo:", self.fenotipo)
        print("Fitness:", self.fitness)

    def mutar(self):
        print(f"El individuo {self.nombre} tuvo una mutación simple en sus genes.")

    def evaluar_fitness(self):
        print(f"Se evalúa la aptitud de {self.nombre}.")
```

```python
ind1 = Individuo("Individuo A", ["A", "T", "G", "C"])
ind2 = Individuo("Individuo B", ["A", "A", "G", "T"])

poblacion1 = Poblacion([ind1, ind2])

poblacion1.mostrar_poblacion()
poblacion1.evaluar_poblacion()
poblacion1.reproducir()
poblacion1.evolucionar()
```
