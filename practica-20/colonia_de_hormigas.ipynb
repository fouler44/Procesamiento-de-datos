## Problema del viajero

### Descripción general
El problema del viajero consiste en encontrar una ruta para pasar por varias ciudades sin repetir ninguna y regresar al punto de inicio, tratando de que la distancia total sea la menor posible.

La idea es simple, pero cuando hay muchas ciudades se vuelve complicado, porque hay muchísimas formas de acomodar el recorrido.


### Objetivos

- Encontrar la ruta más corta  
- Ahorrar tiempo o recursos (como gasolina)  
- Obtener una solución que sea buena, aunque no siempre sea perfecta  


### Limitaciones

- Hay demasiadas combinaciones posibles  
- Encontrar la mejor solución exacta puede tardar mucho  
- No siempre es viable resolverlo de forma exacta  
- Muchas veces se usan métodos que dan soluciones cercanas  


### Trabajos previos
Se han propuesto varios métodos para resolver este problema. Uno que se menciona es:

**Algoritmo de colonias de hormigas**  
  Está basado en cómo las hormigas encuentran caminos cortos usando feromonas. Con el tiempo, las mejores rutas se refuerzan y se vuelven más probables.

En general, hay dos formas de resolverlo:

- Métodos exactos
- Métodos heurísticos

Por eso normalmente se prefieren los heurísticos, porque dan resultados aceptables en menos tiempo.

```python
import random, sys, math

# Genera una matriz de distancias
def matrizDistancias(nCiud, distanciaMaxima):
    matriz = [[0 for i in range(nCiud)] for j in range(nCiud)]
    for i in range(nCiud):
        for j in range(i):
            matriz[i][j] = distanciaMaxima * random.random()
            matriz[j][i] = matriz[i][j]
    return matriz

# Elige la siguiente ciudad
def eligeCiudad(dists, ferom, visitadas):
    listaPesos = []
    disponibles = []
    actual = visitadas[-1]

    alfa = 1.0
    beta = 0.5

    for i in range(len(dists)):
        if i not in visitadas:
            fer = math.pow((1.0 + ferom[actual][i]), alfa)
            peso = math.pow(1.0 / dists[actual][i], beta) * fer
            disponibles.append(i)
            listaPesos.append(peso)

    valor = random.random() * sum(listaPesos)
    acumulado = 0.0
    i = -1

    while valor > acumulado:
        i += 1
        acumulado += listaPesos[i]

    return disponibles[i]

# Genera un camino
def eligeCamino(distancias, feromonas):
    camino = [0]
    longCamino = 0

    while len(camino) < len(distancias):
        ciudad = eligeCiudad(distancias, feromonas, camino)
        longCamino += distancias[camino[-1]][ciudad]
        camino.append(ciudad)

    longCamino += distancias[camino[-1]][0]
    camino.append(0)

    return (camino, longCamino)

# Actualiza feromonas
def rastroFeromonas(feromonas, camino, dosis):
    for i in range(len(camino) - 1):
        feromonas[camino[i]][camino[i + 1]] += dosis

# Evaporación de feromonas
def evaporaFeromonas(feromonas):
    for lista in feromonas:
        for i in range(len(lista)):
            lista[i] *= 0.9

# Algoritmo principal
def hormigas(distancias, iteraciones, distMedia):
    n = len(distancias)
    feromonas = [[0 for i in range(n)] for j in range(n)]

    mejorCamino = []
    longMejorCamino = float("inf")

    for _ in range(iteraciones):
        camino, longCamino = eligeCamino(distancias, feromonas)

        if longCamino <= longMejorCamino:
            mejorCamino = camino
            longMejorCamino = longCamino

            rastroFeromonas(feromonas, camino, distMedia / longCamino)

        evaporaFeromonas(feromonas)

    return (mejorCamino, longMejorCamino)

# Prueba
numCiudades = 10
distanciaMaxima = 10

ciudades = matrizDistancias(numCiudades, distanciaMaxima)

iteraciones = 1000
distMedia = numCiudades * distanciaMaxima / 2

camino, longCamino = hormigas(ciudades, iteraciones, distMedia)

print("Camino:", camino)
print("Longitud del camino:", longCamino)
```
