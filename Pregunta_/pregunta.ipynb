```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Leer los datos

```python
df = pd.read_csv('hotel.csv')
```

### Gráfica de corelación

```python
plt.figure()
sns.regplot(x="Hab_Ocupadas", y="Ingreso", data=df)
plt.xlabel("Habitaciones Ocupadas")
plt.ylabel("Ingreso del Restaurante")
plt.title("Correlación con Línea de Regresión")
plt.show()
```

En la gráfica podemos observar una correlación positiva, pero también podemos ver puntos que no están alineados por lo que no es una correlación fuerte, es más moderada pues siguen el camino de la línea pero hay un poco de dispersión.

```python
correlacion = df["Hab_Ocupadas"].corr(df["Ingreso"])
correlacion
```

### Conclusión

El coeficiente de correlación fue r = 0.4229, esto nos indica que si hay una relación positiva entre el número de las habitaciones ocupadas y el ingreso, pero no es super fuerte, es moderada. Por lo que generalmente el ingreso subirá, pero no es infalible
