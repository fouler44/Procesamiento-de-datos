# Convertir grados Celsius a grados Fahrenheit

### Paso 1. Importar las librerías

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

### Paso 2. Obtención de los datos

```python
celsius = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)
fahrenheit = np.array([-40, 14, 32, 46, 59, 72, 100], dtype=float)
```

```python
plt.scatter(celsius, fahrenheit)

plt.title('Correlación entre Celsius y Fahrenheit')
plt.xlabel('Celsius')
plt.ylabel('Fahrenheit')

plt.show()
```

### Paso 3. Selección de las variables independiente y objetivo

```python
X_train = celsius
y_train = fahrenheit
```

### Paso 4. Creación del modelo

```python
print("Creando el modelo")
model = tf.keras.Sequential()
model.add(tf.keras.layers.Input(shape=(1,)))
model.add(tf.keras.layers.Dense(units=1))
```

### Paso 5. Compilación del modelo

```python
print("Compilando el modelo")
model.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

### Paso 6. Entrenamiento del modelo

```python
print("Entrenando el modelo")
epochs_hist = model.fit(X_train, y_train, epochs=100)
```

```python
# Evaluación del modelo
print("Evaluando el modelo entrenado")
print("Keys: ")
print(epochs_hist.history.keys())

# Gráfico
plt.plot(epochs_hist.history['loss'])
plt.title("Progreso de pérdida durante el entrenamiento del modelo")
plt.xlabel("Epochs")
plt.ylabel("Training loss")
plt.legend(["Training loss"])
plt.show()
```

```python
Temp_C = 100
Temp_C_array = np.array([Temp_C])
Temp_F = model.predict(Temp_C_array)

print(f"Temperatura de predicción: {str(Temp_F)}")
```

```python
Temp_F = 9/5 * Temp_C + 32
print(f"Temperatura de Ecuación: {str(Temp_F)}")
```

### __1. El optimizador de algoritmo utilizado es Adam, pero existen otros como SGD y RMSProp, describe cada uno de ello y explica como pudieran cambiar los resultados de tu modelo.__
El optimizador Adam es un algoritmo que ajusta los pesos de la red neuronal combinando dos ideas: el promedio de los gradientes y la velocidad con la que cambian. Esto permite que el entrenamiento sea más rápido y estable, por lo que suele funcionar bien en muchos modelos.

El optimizador SGD (Stochastic Gradient Descent) es uno de los métodos más simples. Ajusta los pesos usando el gradiente calculado en cada iteración. Aunque es más sencillo, puede ser más lento para converger y a veces puede quedarse en mínimos locales.

Por otro lado, RMSProp ajusta la tasa de aprendizaje dependiendo del comportamiento reciente de los gradientes. Esto ayuda a que el entrenamiento sea más estable, especialmente cuando los datos tienen variaciones grandes.

Si cambiamos el optimizador, los resultados del modelo podrían variar en la velocidad de entrenamiento y la precisión final. Algunos optimizadores pueden tardar más en aprender o requerir más épocas para llegar a un buen resultado.

### __2. Describe porque es importante agregar una función de pérdida como 'mean_squared_error ', investiga que otra función de pérdida se puede implementar y como alteraría los resultados.__

La función de pérdida mean_squared_error es importante porque mide qué tan lejos están las predicciones del modelo respecto a los valores reales. Básicamente calcula el promedio del cuadrado del error, lo que permite al modelo ajustar sus parámetros para minimizar ese error.

### __3. En las modificaciones realizadas,  cuantas epocas fueron 'suficientes' para que el modelo se considere entrenado para ponerlo en funcionamiento.__
300 epochs fueron suficientes.

### __4. ¿Por qué son importantes las epoch?__
Las epoch representan el número de veces que el modelo recorre todo el conjunto de datos durante el entrenamiento. Son importantes porque permiten que el modelo aprenda gradualmente y mejore sus predicciones.

Si hay muy pocas epochs, el modelo no aprende lo suficiente.<br>
Si hay demasiadas, el modelo puede sobreajustarse a los datos.

### __5. Que función de activación utiliza el modelo? ¿Es posible cambiarla? Si es así, cambia a una función RELU y muestra el resultado.__
El modelo original utiliza la función de activación lineal, y sí, es posible cambiarla.

```python
model_2 = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1)
])
```

```python
model_2 = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1)
])

model_2.compile(
    optimizer=tf.keras.optimizers.Adam(1.0),
    loss='mean_squared_error'
)

history_2_relu = model_2.fit(X_train, y_train, epochs=300, verbose=0)
```

```python
plt.figure()
plt.plot(history_2_relu.history['loss'])
plt.title("Loss (MSE) - model_2 con ReLU")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

La función ReLU devuelve 0 cuando el valor es negativo y mantiene el valor cuando es positivo. Esto ayuda a que la red neuronal pueda aprender relaciones más complejas.
